## 1. LATAR BELAKANG

Paparan panas di lingkungan kerja merupakan salah satu faktor risiko yang dapat memengaruhi kesehatan dan produktivitas pekerja. Paparan panas yang berlebihan dapat meningkatkan beban termal tubuh dan menyebabkan gangguan kesehatan selama maupun setelah bekerja. Flouris dkk. (2018) melakukan tinjauan sistematis dan meta-analisis terhadap penelitian tekanan panas akibat pekerjaan. Hasil analisis tersebut menunjukkan bahwa sekitar 35% pekerja yang mengalami paparan panas mengalami tekanan panas akibat pekerjaan, sedangkan sekitar 30% pekerja melaporkan penurunan produktivitas. Penelitian tersebut juga menemukan bahwa pekerja yang menjalani satu giliran kerja dalam kondisi tekanan panas memiliki kemungkinan mengalami tekanan panas sekitar empat kali lebih tinggi daripada pekerja dalam kondisi termonetral. Temuan tersebut menunjukkan bahwa paparan panas memerlukan perhatian dalam pengelolaan keselamatan dan kesehatan kerja.

Penilaian tekanan panas di lingkungan kerja memerlukan indikator yang dapat merepresentasikan kondisi termal berdasarkan beberapa parameter lingkungan. Salah satu indikator yang digunakan secara luas adalah *Wet Bulb Globe Temperature* (WBGT), yang dalam peraturan Indonesia disebut Indeks Suhu Basah dan Bola (ISBB). Standar ISO 7243:2017 menetapkan WBGT sebagai metode penyaringan untuk menilai tekanan panas yang dialami seseorang selama paparan kerja, baik di dalam maupun di luar ruangan, dengan cakupan penilaian hingga delapan jam kerja (International Organization for Standardization [ISO], 2017). Peraturan Menteri Ketenagakerjaan Republik Indonesia Nomor 5 Tahun 2018 menggunakan ISBB sebagai parameter penilaian iklim kerja panas. Peraturan tersebut mendefinisikan iklim kerja sebagai perpaduan suhu, kelembapan, kecepatan gerakan udara, panas radiasi, dan pengeluaran panas tubuh akibat pekerjaan (Kementerian Ketenagakerjaan Republik Indonesia, 2018).

Pemerintah Indonesia menetapkan ketentuan mengenai nilai ambang batas iklim kerja melalui Peraturan Menteri Ketenagakerjaan Republik Indonesia Nomor 5 Tahun 2018, khususnya Lampiran 1.A. Tabel tersebut menghubungkan empat kategori beban kerja, yaitu ringan, sedang, berat, dan sangat berat, dengan empat interval pengaturan waktu kerja setiap jam, yaitu 75–100%, 50–75%, 25–50%, dan 0–25%. Sebagai contoh, tabel tersebut menetapkan ambang ISBB sebesar 31,0 °C untuk beban kerja ringan dan 28,0 °C untuk beban kerja sedang pada pengaturan waktu kerja 75–100%. Pada pengaturan waktu kerja 0–25%, tabel tersebut menetapkan ambang sebesar 32,5 °C untuk beban kerja ringan, 31,5 °C untuk beban kerja sedang, 30,5 °C untuk beban kerja berat, dan 30,0 °C untuk beban kerja sangat berat. Peraturan tersebut juga mewajibkan pengendalian apabila hasil pengukuran melampaui nilai ambang batas, termasuk melalui pengaturan atau pembatasan waktu pajanan pekerja (Kementerian Ketenagakerjaan Republik Indonesia, 2018, Pasal 9 dan Lampiran 1.A).

Penelitian mengenai pengaturan waktu kerja dan istirahat di lingkungan panas telah membahas penggunaan berbagai regimen untuk mengendalikan paparan panas pada pekerja. Deshayes dkk. (2024) melakukan tinjauan cakupan terhadap 49 publikasi mengenai regimen kerja-istirahat di lingkungan panas, yang terdiri atas 35 penelitian eksperimental dan 14 sumber lainnya. Hasil tinjauan tersebut menunjukkan bahwa sebagian besar penelitian eksperimental berlangsung di laboratorium dan melibatkan subjek yang didominasi laki-laki. Tinjauan tersebut juga menemukan bahwa 23 dari 35 penelitian eksperimental atau sekitar 66% tidak menjelaskan dasar penetapan regimen kerja-istirahat yang digunakan. Selain itu, 27 dari 35 penelitian eksperimental atau sekitar 77% melaporkan bahwa suhu inti tubuh peserta melampaui 38 °C pada kondisi eksperimen tertentu. Temuan tersebut menunjukkan bahwa penelitian mengenai pengaturan kerja-istirahat masih memerlukan penjelasan yang transparan mengenai dasar pembentukan aturan serta evaluasi karakteristik penerapannya.

Perkembangan teknologi pengukuran lingkungan memungkinkan peneliti memperoleh data WBGT melalui pengujian laboratorium dan pengamatan lapangan. Ahmadinejad (2026) memublikasikan kumpulan data berjudul *Design, Development, and Performance Evaluation of a Low-Cost IoT-Based System for Real-Time Occupational Heat Stress Monitoring* melalui Mendeley Data. Kumpulan data tersebut memuat data eksperimen dan lapangan untuk mengevaluasi sensor serta sistem pemantauan tekanan panas berbasis Internet of Things. Data tersebut mencakup pengukuran deret waktu berupa suhu, kelembapan relatif, suhu bola, dan WBGT yang dikumpulkan menggunakan prototipe berbasis Internet of Things serta perangkat referensi. Data ini menyediakan pengukuran lingkungan untuk pengujian komputasional, tetapi tidak menyediakan label beban kerja aktual dan catatan perilaku kerja-istirahat setiap pekerja (Ahmadinejad, 2026).

Berdasarkan fakta tersebut, gagasan ilmiah ini mengusulkan sistem pendukung keputusan berbasis Fuzzy Mamdani untuk menghasilkan rekomendasi proporsi waktu kerja dan istirahat berdasarkan ISBB serta kategori beban kerja. Pemilihan logika fuzzy didasarkan pada kemampuannya merepresentasikan derajat keanggotaan secara bertahap pada rentang 0 sampai 1 (Zadeh, 1965), sedangkan metode Mamdani memungkinkan sistem mengolah masukan melalui aturan linguistik berbentuk JIKA–MAKA (Mamdani & Assilian, 1975). Penelitian ini menyusun fungsi keanggotaan dan basis aturan dengan mengacu pada nilai ambang numerik dalam Lampiran 1.A Permenaker Nomor 5 Tahun 2018. Model menghasilkan persentase waktu kerja sebagai keluaran utama dan menghitung persentase waktu istirahat sebagai komplemennya. Penelitian mengevaluasi karakteristik keluaran Fuzzy Mamdani melalui perbandingan dengan pembacaan langsung tabel regulasi, pengujian konsistensi matematis, dan analisis sensitivitas terhadap perubahan nilai ISBB. Nilai WBGT lapangan digunakan sebagai masukan lingkungan, sedangkan empat kategori beban kerja diperlakukan sebagai skenario komputasional. Dengan demikian, penelitian ini berfokus pada penerapan dan evaluasi perilaku model berbasis aturan yang transparan serta tidak mengklaim validasi fisiologis atau pembuktian efektivitas pengaturan kerja-istirahat pada pekerja secara langsung.

````python
# XLSX juga merupakan arsip ZIP secara internal; periksa ekstensi sebelum mendeteksi ZIP Mendeley.
if nama_upload.lower().endswith(".xlsx"):
    nama_xlsx = nama_upload
    bytes_xlsx = bytes_upload
elif zipfile.is_zipfile(io.BytesIO(bytes_upload)):
    with zipfile.ZipFile(io.BytesIO(bytes_upload)) as zf:
        daftar_xlsx = [
            n for n in zf.namelist()
            if n.lower().endswith(".xlsx") and not n.startswith("__MACOSX/")
        ]
        if len(daftar_xlsx) != 1:
            raise RuntimeError(
                f"Diharapkan tepat satu workbook XLSX di dalam ZIP; ditemukan {len(daftar_xlsx)}: {daftar_xlsx}"
            )
        nama_xlsx = daftar_xlsx[0]
        bytes_xlsx = zf.read(nama_xlsx)
else:
    raise ValueError("Berkas harus berupa ZIP Mendeley atau Dataset.xlsx.")

sha256_xlsx = hashlib.sha256(bytes_xlsx).hexdigest()
print("Workbook:", nama_xlsx)
print("Ukuran XLSX:", f"{len(bytes_xlsx):,}", "byte")
print("SHA-256 XLSX:", sha256_xlsx)

xls = pd.ExcelFile(io.BytesIO(bytes_xlsx), engine="openpyxl")
print("\nDaftar worksheet:")
for nama in xls.sheet_names:
    df_temp = pd.read_excel(xls, sheet_name=nama)
    print(f"- {nama}: {len(df_temp)} baris data × {len(df_temp.columns)} kolom")

sheet_wajib = [
    "Final Field Test outdoor", "Final Field Test indoor",
    "Final Lab Test 1", "Final Lab Test 2", "Final Lab Test 3"
]
normalisasi_sheet = lambda teks: re.sub(r"[^a-z0-9]", "", str(teks).lower())
peta_sheet = {normalisasi_sheet(nama): nama for nama in xls.sheet_names}
sheet_hilang = [s for s in sheet_wajib if normalisasi_sheet(s) not in peta_sheet]
if sheet_hilang:
    raise RuntimeError(f"Worksheet wajib tidak ditemukan: {sheet_hilang}. Sheet tersedia: {xls.sheet_names}")

workbook_asli = pd.read_excel(xls, sheet_name=None)
workbook = {nama_kanonis: workbook_asli[peta_sheet[normalisasi_sheet(nama_kanonis)]] for nama_kanonis in sheet_wajib}
print("\nWorkbook berhasil dibaca. Nama sheet kanonis:")
for nama in workbook:
    print(f"- {nama}: {len(workbook[nama])} baris")
````

## Audit struktur data lapangan

````python
def normalisasi_kolom(teks):
    return re.sub(r"[^a-z0-9]", "", str(teks).lower())


def cari_kolom(df, nama_target):
    target = normalisasi_kolom(nama_target)
    peta = {normalisasi_kolom(c): c for c in df.columns}
    if target not in peta:
        raise KeyError(
            f"Kolom '{nama_target}' tidak ditemukan. Kolom tersedia: {list(df.columns)}"
        )
    return peta[target]


KONFIG_LAPANGAN = {
    "Final Field Test outdoor": {
        "sumber": "Outdoor",
        "kolom_wbgt": "WBGT outdoor",
        "kolom_twb": "Wet BulbTemperature",
        "kolom_tg": "Globe Temperature",
        "kolom_ta": "Temperature",
        "kolom_rh": "Relative Humidity",
    },
    "Final Field Test indoor": {
        "sumber": "Indoor",
        "kolom_wbgt": "WBGT Indoor",
        "kolom_twb": "Wet BulbTemperature",
        "kolom_tg": "Globe Temperature",
        "kolom_ta": "Temperature",
        "kolom_rh": "Relative Humidity",
    },
}

baris_lapangan = []
ringkasan_struktur = []

for nama_sheet, cfg in KONFIG_LAPANGAN.items():
    df = workbook[nama_sheet].copy()
    kolom = {
        "waktu": cari_kolom(df, "Time"),
        "wbgt": cari_kolom(df, cfg["kolom_wbgt"]),
        "twb": cari_kolom(df, cfg["kolom_twb"]),
        "tg": cari_kolom(df, cfg["kolom_tg"]),
        "ta": cari_kolom(df, cfg["kolom_ta"]),
        "rh": cari_kolom(df, cfg["kolom_rh"]),
        "di": cari_kolom(df, "DI"),
    }
    df_out = pd.DataFrame({
        "Sumber": cfg["sumber"],
        "Worksheet": nama_sheet,
        "Baris_Excel": np.arange(len(df)) + 2,
        "Waktu": df[kolom["waktu"]],
        "Suhu_udara_C": pd.to_numeric(df[kolom["ta"]], errors="coerce"),
        "RH_persen": pd.to_numeric(df[kolom["rh"]], errors="coerce"),
        "Suhu_basah_alami_C": pd.to_numeric(df[kolom["twb"]], errors="coerce"),
        "Suhu_bola_C": pd.to_numeric(df[kolom["tg"]], errors="coerce"),
        "WBGT_tercatat_C": pd.to_numeric(df[kolom["wbgt"]], errors="coerce"),
        "DI": pd.to_numeric(df[kolom["di"]], errors="coerce"),
    })
    baris_lapangan.append(df_out)
    ringkasan_struktur.append({
        "Sumber": cfg["sumber"],
        "Worksheet": nama_sheet,
        "Baris": len(df),
        "Kolom_WBGT": kolom["wbgt"],
        "WBGT_kosong": int(df_out["WBGT_tercatat_C"].isna().sum()),
        "WBGT_min_C": df_out["WBGT_tercatat_C"].min(),
        "WBGT_maks_C": df_out["WBGT_tercatat_C"].max(),
    })

df_lapangan = pd.concat(baris_lapangan, ignore_index=True)
df_struktur = pd.DataFrame(ringkasan_struktur)

display(df_struktur.style.format({"WBGT_min_C": "{:.3f}", "WBGT_maks_C": "{:.3f}"}))
print("Total baris lapangan:", len(df_lapangan))
print("WBGT kosong:", int(df_lapangan["WBGT_tercatat_C"].isna().sum()))
print("\nPreview data aktual:")
display(df_lapangan.head(8))
````

## Audit asal pengukuran WBGT lapangan

Metadata Mendeley menyatakan bahwa dataset memuat WBGT dari prototipe IoT dan perangkat referensi. Namun, asal setiap kolom WBGT pada worksheet lapangan harus ditentukan dari struktur workbook, bukan diasumsikan dari deskripsi umum dataset.

Kode berikut memeriksa apakah worksheet field test memiliki kolom WBGT terpisah yang secara eksplisit dinamai `Reference`, `Developed System`, `Prototype`, atau sejenisnya. Jika hanya terdapat satu kolom WBGT tanpa identitas instrumen, asal instrumennya dinyatakan **belum teridentifikasi**. Dalam kondisi itu, jangan menyebut kolom tersebut sebagai nilai prototipe atau referensi secara pasti.

## Pemeriksaan timestamp dan rentang data

Sheet lapangan menyediakan kolom `Time`, tetapi tidak menyediakan tanggal atau ID pekerja. Karena itu, notebook hanya memeriksa interval sampling dalam setiap sheet. Data ini **tidak** dianggap sebagai sampel independen dan tidak digunakan untuk mengklaim paparan sepanjang shift 8 jam.

````python
def waktu_ke_detik(nilai):
    if pd.isna(nilai):
        return np.nan
    if isinstance(nilai, pd.Timestamp):
        nilai = nilai.time()
    if isinstance(nilai, (datetime, time)):
        return nilai.hour * 3600 + nilai.minute * 60 + nilai.second + nilai.microsecond / 1e6
    if isinstance(nilai, (int, float, np.integer, np.floating)):
        return float(nilai) * 86400.0
    teks = str(nilai).strip()
    try:
        bagian = teks.split(":")
        if len(bagian) >= 3:
            return int(bagian[0]) * 3600 + int(bagian[1]) * 60 + float(bagian[2])
    except Exception:
        return np.nan
    return np.nan


ringkasan_waktu = []
for nama_sheet, cfg in KONFIG_LAPANGAN.items():
    waktu = workbook[nama_sheet][cari_kolom(workbook[nama_sheet], "Time")].map(waktu_ke_detik)
    waktu = waktu.dropna().to_numpy(dtype=float)
    delta = np.diff(waktu)
    delta = np.where(delta < 0, delta + 86400.0, delta)
    delta = delta[delta > 0]
    ringkasan_waktu.append({
        "Sumber": cfg["sumber"],
        "Jumlah_timestamp": len(waktu),
        "Interval_median_detik": float(np.median(delta)) if len(delta) else np.nan,
        "Interval_min_detik": float(np.min(delta)) if len(delta) else np.nan,
        "Interval_maks_detik": float(np.max(delta)) if len(delta) else np.nan,
    })
display(pd.DataFrame(ringkasan_waktu).style.format({
    "Interval_median_detik": "{:.1f}",
    "Interval_min_detik": "{:.1f}",
    "Interval_maks_detik": "{:.1f}",
}))

print("Interpretasi: timestamp adalah waktu pencatatan dalam rangkaian pengujian; bukan tanggal, identitas pekerja, atau bukti cakupan shift penuh.")
````

## Definisi model Fuzzy Mamdani

````python
# Sembilan MF input ISBB
MF_ISBB = {
    "T27.5": (15.0, 15.0, 27.5, 28.0),
    "T28":   (27.5, 28.0, 28.0, 29.0),
    "T29":   (28.0, 29.0, 29.0, 30.0),
    "T30":   (29.0, 30.0, 30.0, 30.5),
    "T30.5": (30.0, 30.5, 30.5, 31.0),
    "T31":   (30.5, 31.0, 31.0, 31.5),
    "T31.5": (31.0, 31.5, 31.5, 32.0),
    "T32":   (31.5, 32.0, 32.0, 32.5),
    "T32.5": (32.0, 32.5, 32.5, 32.5),
}

KATEGORI_BEBAN = {1: "Ringan", 2: "Sedang", 3: "Berat", 4: "Sangat berat"}
MF_OUTPUT = {
    "P0-25":   (-12.5, 0.0, 25.0, 37.5),
    "P25-50":  (12.5, 25.0, 50.0, 62.5),
    "P50-75":  (37.5, 50.0, 75.0, 87.5),
    "P75-100": (62.5, 75.0, 100.0, 112.5),
}

MATRIKS_RULE = {
    "T27.5": ["P75-100", "P75-100", "P50-75", "P25-50"],
    "T28":   ["P75-100", "P75-100", "P25-50", "P25-50"],
    "T29":   ["P75-100", "P50-75", "P25-50", "P0-25"],
    "T30":   ["P75-100", "P25-50", "P0-25", "P0-25"],
    "T30.5": ["P75-100", "P0-25", "P0-25", None],
    "T31":   ["P75-100", "P0-25", None, None],
    "T31.5": ["P25-50", "P0-25", None, None],
    "T32":   ["P25-50", None, None, None],
    "T32.5": ["P0-25", None, None, None],
}

AMBANG_REGULASI = {
    1: [("P75-100", 31.0), ("P50-75", 31.0), ("P25-50", 32.0), ("P0-25", 32.5)],
    2: [("P75-100", 28.0), ("P50-75", 29.0), ("P25-50", 30.0), ("P0-25", 31.5)],
    3: [("P75-100", None), ("P50-75", 27.5), ("P25-50", 29.0), ("P0-25", 30.5)],
    4: [("P75-100", None), ("P50-75", None), ("P25-50", 28.0), ("P0-25", 30.0)],
}

BATAS_ATAS_REGIMEN = {"P75-100": 100.0, "P50-75": 75.0, "P25-50": 50.0, "P0-25": 25.0}
DOMAIN_ISBB = (15.0, 32.5)
DOMAIN_OUTPUT = (-12.5, 112.5)
EPS_AMBANG = 1e-9

RULES = []
for nama_mf, daftar in MATRIKS_RULE.items():
    for kode, konsekuen in enumerate(daftar, start=1):
        if konsekuen is not None:
            RULES.append({
                "mf_isbb": nama_mf,
                "kode_beban": kode,
                "nama_beban": KATEGORI_BEBAN[kode],
                "konsekuen": konsekuen,
            })

assert len(RULES) == 25
print("Jumlah aturan:", len(RULES))
print("Jumlah kombinasi n/a:", 36 - len(RULES))
````

## Fungsi keanggotaan, centroid, lookup regulasi, dan inferensi

````python
def trapmf_scalar(x, params):
    a, b, c, d = map(float, params)
    x = float(x)
    if x < a or x > d:
        return 0.0
    if b <= x <= c:
        return 1.0
    if x < b:
        return (x - a) / (b - a) if b > a else 1.0
    if x > c:
        return (d - x) / (d - c) if d > c else 1.0
    return 0.0


def trapmf_array(x, params):
    a, b, c, d = map(float, params)
    x = np.asarray(x, dtype=float)
    mu = np.zeros_like(x, dtype=float)
    if b > a:
        m = (x > a) & (x < b)
        mu[m] = (x[m] - a) / (b - a)
    mu[(x >= b) & (x <= c)] = 1.0
    if d > c:
        m = (x > c) & (x < d)
        mu[m] = (d - x[m]) / (d - c)
    return mu


def titik_potong_clipped(params, alpha):
    a, b, c, d = map(float, params)
    titik = sorted(set([a, b, c, d]))
    hasil = list(titik)
    for x0, x1 in zip(titik[:-1], titik[1:]):
        m0, m1 = trapmf_scalar(x0, params), trapmf_scalar(x1, params)
        if (m0 - alpha) * (m1 - alpha) < 0 and abs(m1 - m0) > 1e-15:
            xp = x0 + (alpha - m0) * (x1 - x0) / (m1 - m0)
            if x0 + 1e-14 < xp < x1 - 1e-14:
                hasil.append(float(xp))
    return sorted(set(hasil))


def centroid_agregasi_eksak(terms):
    alpha_per_mf = {}
    for nama, alpha in terms:
        if alpha > 0:
            alpha_per_mf[nama] = max(alpha_per_mf.get(nama, 0.0), float(alpha))
    aktif = list(alpha_per_mf.items())
    if not aktif:
        return {"luas": 0.0, "momen": 0.0, "centroid": np.nan}
    if len(aktif) > 2:
        raise ValueError("Lebih dari dua konsekuen aktif; periksa partisi input.")

    if len(aktif) == 1:
        nama, alpha = aktif[0]
        params = MF_OUTPUT[nama]
        titik = titik_potong_clipped(params, alpha)
        fungsi = lambda x: min(alpha, trapmf_scalar(x, params))
    else:
        (nama1, alpha1), (nama2, alpha2) = aktif
        params1, params2 = MF_OUTPUT[nama1], MF_OUTPUT[nama2]
        f1 = lambda x: min(alpha1, trapmf_scalar(x, params1))
        f2 = lambda x: min(alpha2, trapmf_scalar(x, params2))
        titik = sorted(set(
            titik_potong_clipped(params1, alpha1)
            + titik_potong_clipped(params2, alpha2)
        ))
        silang = []
        for x0, x1 in zip(titik[:-1], titik[1:]):
            d0, d1 = f1(x0) - f2(x0), f1(x1) - f2(x1)
            if d0 * d1 < -1e-20 and abs(d1 - d0) > 1e-15:
                xc = x0 - d0 * (x1 - x0) / (d1 - d0)
                if x0 + 1e-13 < xc < x1 - 1e-13:
                    silang.append(float(xc))
        titik = sorted(set(titik + silang))
        fungsi = lambda x: max(f1(x), f2(x))

    nilai = [fungsi(x) for x in titik]
    luas = 0.0
    momen = 0.0
    for x0, x1, y0, y1 in zip(titik[:-1], titik[1:], nilai[:-1], nilai[1:]):
        dx = x1 - x0
        luas += dx * (y0 + y1) / 2.0
        momen += dx * ((2*x0 + x1)*y0 + (x0 + 2*x1)*y1) / 6.0
    centroid = momen / luas if luas > 0 else np.nan
    return {"luas": float(luas), "momen": float(momen), "centroid": float(centroid)}


def validasi_dan_lookup(isbb, kode_beban, eps=EPS_AMBANG):
    if isinstance(isbb, (bool, np.bool_)) or not isinstance(isbb, (int, float, np.integer, np.floating)):
        return {"status": "INPUT_TIDAK_VALID", "pesan": "ISBB harus berupa angka."}
    isbb = float(isbb)
    if not np.isfinite(isbb):
        return {"status": "INPUT_TIDAK_VALID", "pesan": "ISBB harus berupa bilangan hingga."}
    if isinstance(kode_beban, (bool, np.bool_)) or not isinstance(kode_beban, (int, np.integer)):
        return {"status": "INPUT_TIDAK_VALID", "pesan": "Kode beban harus bilangan bulat 1–4."}
    kode_beban = int(kode_beban)
    if kode_beban not in KATEGORI_BEBAN:
        return {"status": "INPUT_TIDAK_VALID", "pesan": "Kategori beban tidak dikenal."}
    if isbb < DOMAIN_ISBB[0]:
        return {"status": "DI_LUAR_DOMAIN_KOMPUTASI", "pesan": "Di luar domain komputasi model."}

    for regimen, ambang in AMBANG_REGULASI[kode_beban]:
        if ambang is not None and isbb <= ambang + eps:
            isbb_fis = min(isbb, float(ambang)) if isbb > ambang else isbb
            return {
                "status": "VALID", "isbb": isbb, "isbb_fis": isbb_fis,
                "kode_beban": kode_beban, "nama_beban": KATEGORI_BEBAN[kode_beban],
                "regimen_acuan": regimen, "ambang_acuan": float(ambang),
                "U": BATAS_ATAS_REGIMEN[regimen],
            }
    return {
        "status": "MELEBIHI_SELURUH_AMBANG_NUMERIK",
        "pesan": "Di luar rentang rekomendasi numerik tabel; diperlukan evaluasi dan pengendalian paparan sesuai ketentuan K3 yang berlaku.",
        "isbb": isbb, "kode_beban": kode_beban,
        "nama_beban": KATEGORI_BEBAN[kode_beban],
    }


def inferensi_fis_saja(isbb, kode_beban):
    mu_t = {nama: trapmf_scalar(isbb, params) for nama, params in MF_ISBB.items()}
    aturan_aktif, terms = [], []
    for nomor, rule in enumerate(RULES, start=1):
        alpha = min(mu_t[rule["mf_isbb"]], 1.0 if rule["kode_beban"] == kode_beban else 0.0)
        if alpha > 0:
            aturan_aktif.append({
                "nomor": nomor, "mf_isbb": rule["mf_isbb"],
                "beban": rule["nama_beban"], "konsekuen": rule["konsekuen"],
                "mu_isbb": mu_t[rule["mf_isbb"]], "alpha": float(alpha),
            })
            terms.append((rule["konsekuen"], float(alpha)))
    hasil = centroid_agregasi_eksak(terms)
    return {"mu_isbb": mu_t, "aturan_aktif": aturan_aktif, "terms": terms, **hasil}


def rekomendasi(isbb, kode_beban):
    info = validasi_dan_lookup(isbb, kode_beban)
    if info["status"] != "VALID":
        return info
    hasil = inferensi_fis_saja(info["isbb_fis"], info["kode_beban"])
    if not np.isfinite(hasil["luas"]) or hasil["luas"] <= 0:
        return {**info, "status": "INFERENSI_TIDAK_TERDEFINISI", "pesan": "Luas agregasi nol atau tidak valid."}
    y_fis = hasil["centroid"]
    if not np.isfinite(y_fis) or not (12.5 - 1e-9 <= y_fis <= 87.5 + 1e-9):
        return {**info, "status": "KESALAHAN_RENTANG_CENTROID", "pesan": "Centroid di luar rentang konfigurasi model."}
    y_final = min(y_fis, info["U"])
    y_final = min(100.0, max(0.0, y_final))
    p_istirahat = 100.0 - y_final
    return {
        **info, **hasil, "status": "VALID", "y_fis": float(y_fis),
        "y_final": float(y_final), "p_istirahat": float(p_istirahat),
        "menit_kerja": float(60*y_final/100),
        "menit_istirahat": float(60*p_istirahat/100),
    }

print("Model Mamdani berhasil dimuat:", len(RULES), "aturan.")
````

````python
df_rekomendasi["Guardrail_aktif"] = (
    df_rekomendasi["Centroid_FIS_persen"].notna()
    & df_rekomendasi["Kerja_final_persen"].notna()
    & ((df_rekomendasi["Centroid_FIS_persen"] - df_rekomendasi["Kerja_final_persen"]).abs() > 1e-9)
)

baris_e2 = []
for (sumber, kode, nama), grup in df_rekomendasi.groupby(
    ["Sumber", "Kode_beban_skenario", "Kategori_beban_skenario"], sort=True
):
    valid = grup[grup["Status"] == "VALID"]
    n = len(grup)
    baris_e2.append({
        "Sumber": sumber,
        "Kode_beban_skenario": kode,
        "Kategori_beban_skenario": nama,
        "N_kombinasi": n,
        "N_valid": len(valid),
        "Persen_valid": 100 * len(valid) / n if n else np.nan,
        "N_melebihi_ambang": int((grup["Status"] == "MELEBIHI_SELURUH_AMBANG_NUMERIK").sum()),
        "N_di_luar_domain": int((grup["Status"] == "DI_LUAR_DOMAIN_KOMPUTASI").sum()),
        "N_input_tidak_valid": int((grup["Status"] == "INPUT_TIDAK_VALID").sum()),
        "N_guardrail_aktif": int(grup["Guardrail_aktif"].sum()),
        "Persen_guardrail_aktif_dari_valid": (
            100 * valid["Guardrail_aktif"].sum() / len(valid) if len(valid) else np.nan
        ),
        "Rerata_centroid_persen": valid["Centroid_FIS_persen"].mean() if len(valid) else np.nan,
        "Rerata_kerja_final_persen": valid["Kerja_final_persen"].mean() if len(valid) else np.nan,
        "Median_kerja_final_persen": valid["Kerja_final_persen"].median() if len(valid) else np.nan,
        "SD_kerja_final_persen": valid["Kerja_final_persen"].std(ddof=1) if len(valid) > 1 else np.nan,
        "Rerata_istirahat_persen": valid["Istirahat_persen"].mean() if len(valid) else np.nan,
    })

df_e2_ringkasan = pd.DataFrame(baris_e2)
display(df_e2_ringkasan.style.format({
    "Persen_valid": "{:.2f}",
    "Persen_guardrail_aktif_dari_valid": "{:.2f}",
    "Rerata_centroid_persen": "{:.2f}",
    "Rerata_kerja_final_persen": "{:.2f}",
    "Median_kerja_final_persen": "{:.2f}",
    "SD_kerja_final_persen": "{:.2f}",
    "Rerata_istirahat_persen": "{:.2f}",
}))
````

## E3 — Perbandingan dengan lookup interval Permenaker

Setiap hasil VALID dibandingkan dengan kategori interval yang diperoleh dari lookup tabel. Output crisp dipetakan menggunakan konvensi endpoint model: `0–25` mencakup 25; `25–50` berarti `25 < p ≤ 50`; `50–75` berarti `50 < p ≤ 75`; dan `75–100` berarti `75 < p ≤ 100`.

Interpretasi: proporsi keluaran pada kategori yang sama, lebih restriktif (persentase kerja lebih rendah), atau lebih permisif. Karena tabel regulasi juga menjadi dasar basis aturan dan guardrail, metrik ini adalah **kesesuaian terhadap acuan desain**, bukan akurasi prediksi independen.

````python
BATAS_INTERVAL_OUTPUT = {
    "P0-25": (0.0, 25.0, 1),
    "P25-50": (25.0, 50.0, 2),
    "P50-75": (50.0, 75.0, 3),
    "P75-100": (75.0, 100.0, 4),
}


def kategori_output(persen):
    if pd.isna(persen):
        return None
    p = float(persen)
    if p <= 25.0 + 1e-9:
        return "P0-25"
    if p <= 50.0 + 1e-9:
        return "P25-50"
    if p <= 75.0 + 1e-9:
        return "P50-75"
    return "P75-100"


hasil_e3 = []
for _, r in df_rekomendasi.iterrows():
    if r["Status"] != "VALID":
        continue
    kelas_fuzzy = kategori_output(r["Kerja_final_persen"])
    kelas_reg = r["Regimen_acuan"]
    ord_fuzzy = BATAS_INTERVAL_OUTPUT[kelas_fuzzy][2]
    ord_reg = BATAS_INTERVAL_OUTPUT[kelas_reg][2]
    lo, hi, _ = BATAS_INTERVAL_OUTPUT[kelas_reg]
    y = float(r["Kerja_final_persen"])
    jarak = max(lo - y, 0.0, y - hi)
    hasil_e3.append({
        "Sumber": r["Sumber"],
        "Worksheet": r["Worksheet"],
        "Baris_Excel": r["Baris_Excel"],
        "Waktu": r["Waktu"],
        "WBGT_tercatat_C": r["WBGT_tercatat_C"],
        "Kategori_beban_skenario": r["Kategori_beban_skenario"],
        "Kategori_lookup_Permenaker": kelas_reg,
        "Kategori_output_fuzzy": kelas_fuzzy,
        "Kerja_final_persen": y,
        "Jarak_ke_interval_pp": jarak,
        "Sama_kategori": kelas_fuzzy == kelas_reg,
        "Lebih_restriktif": ord_fuzzy < ord_reg,
        "Lebih_permisif": ord_fuzzy > ord_reg,
        "Guardrail_aktif": bool(r["Guardrail_aktif"]),
    })

df_e3_detail = pd.DataFrame(hasil_e3)
baris_e3 = []
for (sumber, beban), g in df_e3_detail.groupby(["Sumber", "Kategori_beban_skenario"], sort=True):
    n = len(g)
    baris_e3.append({
        "Sumber": sumber,
        "Kategori_beban_skenario": beban,
        "N_kasus_valid": n,
        "Agreement_rate_persen": 100*g["Sama_kategori"].mean() if n else np.nan,
        "Lebih_restriktif_persen": 100*g["Lebih_restriktif"].mean() if n else np.nan,
        "Lebih_permisif_persen": 100*g["Lebih_permisif"].mean() if n else np.nan,
        "Rerata_jarak_interval_pp": g["Jarak_ke_interval_pp"].mean() if n else np.nan,
        "Guardrail_aktif_persen": 100*g["Guardrail_aktif"].mean() if n else np.nan,
    })

df_e3_ringkasan = pd.DataFrame(baris_e3)
display(df_e3_ringkasan.style.format({c:"{:.2f}" for c in ["Agreement_rate_persen","Lebih_restriktif_persen","Lebih_permisif_persen","Rerata_jarak_interval_pp","Guardrail_aktif_persen"]}))
print("Catatan: lebih-permisif diharapkan 0% karena guardrail membatasi output pada batas atas regimen lookup. Ini bukan bukti keselamatan empiris.")
````

## E4 — Pemeriksaan matematis, batas, dan monotonicity

Pengujian mencakup jumlah aturan, cakupan/partisi fungsi keanggotaan input, validasi kasus manual, input/status khusus, ketepatan transisi ambang, serta monotonicity pada grid WBGT 0,01°C. Hasil grid adalah verifikasi numerik pada resolusi yang dinyatakan, bukan bukti matematis untuk setiap bilangan real.

````python
# E4.1: integritas rule base dan partisi input
assert len(RULES) == 25
assert len(MF_ISBB) == 9
assert set(KATEGORI_BEBAN) == {1, 2, 3, 4}

x_partisi = np.round(np.arange(DOMAIN_ISBB[0], DOMAIN_ISBB[1] + 0.0005, 0.001), 3)
mu_partisi = np.vstack([trapmf_array(x_partisi, p) for p in MF_ISBB.values()]).T
jumlah_mu = mu_partisi.sum(axis=1)
aktif_mu = (mu_partisi > 1e-12).sum(axis=1)

hasil_preflight = {
    "Jumlah_rule": len(RULES),
    "Jumlah_MF_ISBB": len(MF_ISBB),
    "Deviasi_maks_sum_mu_dari_1": float(np.max(np.abs(jumlah_mu - 1.0))),
    "Sum_mu_min": float(jumlah_mu.min()),
    "Sum_mu_max": float(jumlah_mu.max()),
    "Jumlah_titik_dengan_sum_mu_tidak_1": int(np.sum(np.abs(jumlah_mu - 1.0) > 1e-9)),
    "Jumlah_titik_dengan_lebih_dari_2_MF_aktif": int(np.sum(aktif_mu > 2)),
}

# Uji tiga perhitungan manual dan input/status khusus.
kasus_manual_e4 = [(27.5, 1, 87.5), (28.5, 2, 75.0), (30.25, 3, 12.5)]
for t, k, target in kasus_manual_e4:
    z = rekomendasi(t, k)
    assert z["status"] == "VALID" and abs(z["y_final"] - target) <= 1e-6
assert rekomendasi(14.9, 1)["status"] == "DI_LUAR_DOMAIN_KOMPUTASI"
assert rekomendasi(np.nan, 1)["status"] == "INPUT_TIDAK_VALID"
assert rekomendasi(25.0, 9)["status"] == "INPUT_TIDAK_VALID"

# Boundary test untuk setiap ambang numerik pada setiap workload.
baris_boundary = []
for k, daftar in AMBANG_REGULASI.items():
    ambang_unik = sorted({float(a) for _, a in daftar if a is not None})
    for ambang in ambang_unik:
        tepat = validasi_dan_lookup(ambang, k)
        atas = validasi_dan_lookup(ambang + 1e-6, k)
        assert tepat["status"] == "VALID", (k, ambang, tepat)
        baris_boundary.append({
            "Kode_beban": k,
            "Kategori_beban": KATEGORI_BEBAN[k],
            "Ambang_C": ambang,
            "Status_tepat_ambang": tepat["status"],
            "Status_plus_1e_6": atas["status"],
            "Regimen_tepat": tepat.get("regimen_acuan"),
            "Regimen_setelah": atas.get("regimen_acuan"),
        })
df_e4_boundary = pd.DataFrame(baris_boundary)
display(df_e4_boundary)

print("PREFLIGHT MODEL:")
for k, v in hasil_preflight.items():
    print(f"- {k}: {v}")
assert hasil_preflight["Jumlah_titik_dengan_sum_mu_tidak_1"] == 0
assert hasil_preflight["Jumlah_titik_dengan_lebih_dari_2_MF_aktif"] == 0
print("PASS: preflight rule base, partisi input, kasus manual, status, dan ambang.")
````

````python
# E4.2: monotonicity pada grid WBGT 0,01 °C.
# Cache menghindari evaluasi berulang untuk pasangan input yang identik.
from functools import lru_cache

@lru_cache(maxsize=None)
def rekomendasi_cache(x_rounded, k):
    return rekomendasi(float(x_rounded), int(k))

grid_mono = np.round(np.arange(15.0, 32.50001, 0.01), 2)
baris_kurva = []
baris_mono_t = []
for k, nama in KATEGORI_BEBAN.items():
    ys, statuses = [], []
    for t in grid_mono:
        z = rekomendasi_cache(float(t), k)
        y = z.get("y_final") if z.get("status") == "VALID" else np.nan
        ys.append(y)
        statuses.append(z.get("status"))
        baris_kurva.append({"WBGT_C":t,"Kode_beban":k,"Kategori_beban":nama,"Status":z.get("status"),"Kerja_final_persen":y})
    pelanggaran = []
    transisi_ke_status = 0
    transisi_kembali_valid = 0
    for i in range(1, len(grid_mono)):
        if statuses[i-1] == "VALID" and statuses[i] == "VALID":
            delta = ys[i] - ys[i-1]
            if delta > 1e-8:
                pelanggaran.append((grid_mono[i-1], grid_mono[i], ys[i-1], ys[i]))
        elif statuses[i-1] == "VALID" and statuses[i] != "VALID":
            transisi_ke_status += 1
        elif statuses[i-1] != "VALID" and statuses[i] == "VALID":
            transisi_kembali_valid += 1
    baris_mono_t.append({
        "Kategori_beban":nama,
        "N_titik_grid":len(grid_mono),
        "N_pelanggaran_monotonicity_T":len(pelanggaran),
        "N_transisi_VALID_ke_status":transisi_ke_status,
        "N_transisi_status_ke_VALID":transisi_kembali_valid,
        "Contoh_pelanggaran":str(pelanggaran[:3]),
    })

df_e4_kurva = pd.DataFrame(baris_kurva)
df_e4_monotonicity_t = pd.DataFrame(baris_mono_t)

# Monotonicity terhadap beban: untuk WBGT tetap, kenaikan beban tidak boleh meningkatkan kerja.
baris_mono_beban = []
for t in grid_mono:
    out = [rekomendasi_cache(float(t), k) for k in [1,2,3,4]]
    for k in [1,2,3]:
        a, b = out[k-1], out[k]
        y_a = a.get("y_final") if a.get("status") == "VALID" else np.nan
        y_b = b.get("y_final") if b.get("status") == "VALID" else np.nan
        violation = False
        if a.get("status") != "VALID" and b.get("status") == "VALID":
            violation = True
        elif a.get("status") == "VALID" and b.get("status") == "VALID" and y_b > y_a + 1e-8:
            violation = True
        baris_mono_beban.append({
            "WBGT_C":t,"Beban_rendah":KATEGORI_BEBAN[k],"Beban_lebih_tinggi":KATEGORI_BEBAN[k+1],
            "Status_rendah":a.get("status"),"Status_tinggi":b.get("status"),
            "Kerja_rendah_persen":y_a,"Kerja_tinggi_persen":y_b,"Pelanggaran":violation,
        })

df_e4_monotonicity_beban = pd.DataFrame(baris_mono_beban)
ringkas_mono_beban = pd.DataFrame([{
    "Resolusi_grid_C":0.01,
    "N_perbandingan":len(df_e4_monotonicity_beban),
    "N_pelanggaran":int(df_e4_monotonicity_beban["Pelanggaran"].sum()),
    "Persen_pelanggaran":100*df_e4_monotonicity_beban["Pelanggaran"].mean(),
}])

display(df_e4_monotonicity_t)
display(ringkas_mono_beban)
assert df_e4_monotonicity_t["N_pelanggaran_monotonicity_T"].sum() == 0
assert ringkas_mono_beban.loc[0,"N_pelanggaran"] == 0
print(f"PASS: monotonicity numerik pada {len(grid_mono)} titik WBGT, resolusi 0,01 °C.")
print("Peringatan: hasil ini adalah pengujian grid numerik, bukan bukti analitik untuk seluruh bilangan real.")
````

## E5 — Simulasi sensitivitas terhadap perubahan WBGT

E5 menggunakan residual laboratorium `WBGT Developed System − WBGT Reference` pada masing-masing sesi. Untuk setiap sesi dihitung bias (mean residual), kuantil 5%, dan kuantil 95%; ketiganya dipakai sebagai **pergeseran skenario** pada WBGT lapangan, bukan untuk mengoreksi data.

Analisis ini bersyarat: bila asal instrumen WBGT lapangan tidak diketahui, hasil hanya menjawab *seberapa sensitif keluaran model terhadap perubahan WBGT sebesar residual yang teramati di laboratorium*. Hasil tidak membuktikan bahwa error laboratorium sama dengan error lapangan.

````python
# Siapkan distribusi residual tiap sesi laboratorium.
residual_sesi = {}
baris_residual = []
for nama_sheet in ["Final Lab Test 1", "Final Lab Test 2", "Final Lab Test 3"]:
    dfl = workbook[nama_sheet]
    ref = pd.to_numeric(dfl[cari_kolom(dfl, "WBGT Reference")], errors="coerce")
    dev = pd.to_numeric(dfl[cari_kolom(dfl, "WBGT Developed System")], errors="coerce")
    resid = (dev - ref).replace([np.inf, -np.inf], np.nan).dropna().to_numpy(dtype=float)
    if len(resid) == 0:
        continue
    ring = {"Bias_mean":float(np.mean(resid)),"Q05":float(np.quantile(resid,0.05)),"Q95":float(np.quantile(resid,0.95))}
    residual_sesi[nama_sheet] = ring
    baris_residual.append({"Pengujian":nama_sheet,"N_residual":len(resid),**ring,"MAE":float(np.mean(np.abs(resid))),"RMSE":float(np.sqrt(np.mean(resid**2)))})

df_e5_residual = pd.DataFrame(baris_residual)
display(df_e5_residual.style.format({c:"{:.4f}" for c in ["Bias_mean","Q05","Q95","MAE","RMSE"]}))

# Skenario pergeseran: Q05, Bias (rata-rata), dan Q95 dari setiap sesi.
# memo hasil inferensi berdasarkan (WBGT dibulatkan, kategori beban) untuk efisiensi.
@lru_cache(maxsize=None)
def evaluasi_sensitivitas(x_rounded, k):
    z = rekomendasi(float(x_rounded), int(k))
    y = z.get("y_final") if z.get("status") == "VALID" else np.nan
    return z.get("status"), y, z.get("regimen_acuan")

# Lookup baseline dari hasil E2, menggunakan identitas worksheet dan baris Excel.
baseline_map = {}
for _, r in df_rekomendasi.iterrows():
    baseline_map[(r["Worksheet"], int(r["Baris_Excel"]), int(r["Kode_beban_skenario"]))] = (
        r["Status"], r["Kerja_final_persen"], r["Regimen_acuan"]
    )

baris_e5 = []
for nama_sheet, ring in residual_sesi.items():
    skenario_shift = [("Q05", ring["Q05"]), ("Bias_mean", ring["Bias_mean"]), ("Q95", ring["Q95"])]
    for _, row in df_lapangan.iterrows():
        x0 = row["WBGT_tercatat_C"]
        if pd.isna(x0) or not np.isfinite(float(x0)):
            continue
        for k, nama_beban in KATEGORI_BEBAN.items():
            status0, y0, regimen0 = baseline_map[(row["Worksheet"], int(row["Baris_Excel"]), int(k))]
            for label_shift, shift in skenario_shift:
                x1 = float(x0) + float(shift)
                status1, y1, regimen1 = evaluasi_sensitivitas(round(x1, 6), k)
                both_valid = status0 == "VALID" and status1 == "VALID"
                kelas0 = kategori_output(y0) if status0 == "VALID" else None
                kelas1 = kategori_output(y1) if status1 == "VALID" else None
                baris_e5.append({
                    "Sesi_error_laboratorium":nama_sheet,
                    "Skenario_shift":label_shift,
                    "Shift_WBGT_C":float(shift),
                    "Sumber_lapangan":row["Sumber"],
                    "Worksheet_lapangan":row["Worksheet"],
                    "Baris_Excel":int(row["Baris_Excel"]),
                    "Waktu":row["Waktu"],
                    "WBGT_awal_C":float(x0),
                    "WBGT_terganggu_C":x1,
                    "Kode_beban":k,
                    "Kategori_beban_skenario":nama_beban,
                    "Status_awal":status0,
                    "Status_terganggu":status1,
                    "Kerja_awal_persen":y0,
                    "Kerja_terganggu_persen":y1,
                    "Delta_kerja_pp":(float(y1)-float(y0)) if both_valid else np.nan,
                    "Kategori_awal":kelas0,
                    "Kategori_terganggu":kelas1,
                    "Kategori_berubah":(kelas0 != kelas1) if both_valid else np.nan,
                    "Status_berubah":status0 != status1,
                })

df_e5_detail = pd.DataFrame(baris_e5)
baris_e5_ringkas=[]
for (sesi, shift_label, sumber, beban), g in df_e5_detail.groupby(["Sesi_error_laboratorium","Skenario_shift","Sumber_lapangan","Kategori_beban_skenario"], sort=True):
    both = g.dropna(subset=["Delta_kerja_pp"])
    n = len(g)
    baris_e5_ringkas.append({
        "Sesi_error_laboratorium":sesi,
        "Skenario_shift":shift_label,
        "Shift_WBGT_C":g["Shift_WBGT_C"].iloc[0],
        "Sumber_lapangan":sumber,
        "Kategori_beban_skenario":beban,
        "N_kombinasi":n,
        "N_keduanya_VALID":len(both),
        "Status_berubah_persen":100*g["Status_berubah"].mean() if n else np.nan,
        "Kategori_berubah_persen":100*both["Kategori_berubah"].mean() if len(both) else np.nan,
        "Rerata_abs_delta_kerja_pp":both["Delta_kerja_pp"].abs().mean() if len(both) else np.nan,
        "Median_abs_delta_kerja_pp":both["Delta_kerja_pp"].abs().median() if len(both) else np.nan,
        "Maks_abs_delta_kerja_pp":both["Delta_kerja_pp"].abs().max() if len(both) else np.nan,
    })

df_e5_ringkasan = pd.DataFrame(baris_e5_ringkas)
display(df_e5_ringkasan.style.format({c:"{:.2f}" for c in ["Shift_WBGT_C","Status_berubah_persen","Kategori_berubah_persen","Rerata_abs_delta_kerja_pp","Median_abs_delta_kerja_pp","Maks_abs_delta_kerja_pp"]}))
print("Interpretasi E5: perubahan rekomendasi di bawah perturbasi laboratorium adalah simulasi sensitivitas bersyarat, bukan estimasi galat WBGT lapangan atau evaluasi keselamatan.")
````
